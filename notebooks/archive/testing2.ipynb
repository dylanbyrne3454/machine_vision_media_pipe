```python
import os
import pandas as pd
from posedata import PoseData
from tasks.cmj import calculate_jump

folder = "keypoints/cmj"
results = []

for file in os.listdir(folder):

    if file.endswith(".txt"):
        file_path = os.path.join(folder, file)

        try:
            # TXT to pandas dataframe
            pose = PoseData(file_path)

            # Calculate height
            height = calculate_jump(pose.df, fps=30)

            # Add to results
            results.append({
                "File Name": file,
                "Jump Height (cm)" : height
            })
        except Exception as e:
            import traceback
            traceback.print_exc()
            break  # stops at first file so you can see the full output
```

```python
from utilities.utils import flip_axis, get_flight_time, ts_jump_height
import utilities.PTM as PTM
import numpy as np
from posedata import PoseData
from scipy.signal import savgol_filter
# Calculate mm_per_px once from a clean file
reference_file = "keypoints/cmj/P03_CMJBL_FRONT.txt"  # pick your cleanest one
pose_ref = PoseData(reference_file)
df_ref = pose_ref.df.copy()
df_ref[[col for col in df_ref.columns if col.endswith('_x')]] *= 1920
df_ref[[col for col in df_ref.columns if col.endswith('_y')]] *= 1080

hip_x = (df_ref['LEFT_HIP_x'] + df_ref['RIGHT_HIP_x']) / 2
hip_y = (df_ref['LEFT_HIP_y'] + df_ref['RIGHT_HIP_y']) / 2
hip = np.column_stack((hip_x, hip_y))
hip_flipped = flip_axis(hip.copy())
ts = savgol_filter(hip_flipped[:, 1], window_length=11, polyorder=2)

mm_px = PTM.mm_per_px(ts, fps=30, verbose=True, plot=True)
print(f"Using fixed mm_px = {mm_px:.4f} for all files")

# Then use that fixed value in your loop
for file in os.listdir(folder):
    if file.endswith(".txt") and "world" not in file:
        file_path = os.path.join(folder, file)
        try:
            pose = PoseData(file_path)
            df = pose.df.copy()
            df[[col for col in df.columns if col.endswith('_x')]] *= 1920
            df[[col for col in df.columns if col.endswith('_y')]] *= 1080

            height = calculate_jump(df, fps=30, mm_px=mm_px)  # pass it in
            results.append({
                "File Name": file,
                "Jump Height (cm)": height
            })
        except Exception as e:
            import traceback
            traceback.print_exc()
            break
```
